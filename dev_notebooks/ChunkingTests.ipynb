# Code Chunking Development

```python
# Load sample code file: data/manim_sourcecode/manim/mobject/geometry/arc.py
TEST_CODE = ""
with open("../../data/manim_sourcecode/manim/mobject/geometry/arc.py", "r") as file:
    TEST_CODE = file.read()
```

```python
print(TEST_CODE[:400])  # Print first 1000 characters for brevity
```

```python
from synthetic_data_kit.utils.text import split_into_chunks
```

```python
test_chunks = split_into_chunks(TEST_CODE, doc_type='code', chunk_size=10000, overlap=0)
print(f"Number of chunks: {len(test_chunks)}")
for i, chunk in enumerate(test_chunks):
    print(f"Chunk {i+1}:\n{chunk}\n{'-'*40}")
    if i >= 5:  # Limit output to first 5 chunks for brevity
        break
```
